```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../dataset/TTC_LRT_Delays.csv")

avg_delay = df.groupby("Line")["Min Delay"].mean().sort_values()

plt.figure(figsize=(8, 5))
avg_delay.plot(kind="bar")
plt.xlabel("LRT Line")
plt.ylabel("Average Delay (minutes)")
plt.title("Average TTC LRT Delay by Line")
plt.tight_layout()

plt.savefig("viz_1_avg_delay_by_line.png", dpi=300)
plt.show()
```
