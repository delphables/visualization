```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day",
    style="time"
)

plt.title("Tips vs Total Bill by Day and Time")
plt.show()
```

```python
import plotly.graph_objects as go

x = ["A", "B", "C"]
y = [10, 15, 7]

fig = go.Figure()
fig.add_trace(go.Bar(x=x, y=y))
fig.update_layout(title="Simple Plotly Bar Chart")

fig.show()
```
