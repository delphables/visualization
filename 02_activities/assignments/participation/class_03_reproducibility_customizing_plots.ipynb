```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(613)
```

```python
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
ax.legend(loc="lower right")
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")

ax.legend(
    loc="lower right",
    frameon=True,
    fontsize=12,
    ncol=2,
    shadow=True
)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")

# Attach legend's upper-left corner to the plot's top-right corner
ax.legend(loc="upper left", bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc="lower right")

# Place text at an (x, y) position on the axes
ax.text(10, 95, "This value is important", ha="center", color="red", fontsize=12)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc="lower right")

ax.annotate(
    "Important point",
    xy=(10, 95),         # where arrow points (data coords)
    xytext=(30, 94),     # where text sits (data coords)
    arrowprops=dict(arrowstyle="fancy", color="hotpink")
)

plt.show()
```

```python
from matplotlib.ticker import NullLocator, NullFormatter, MaxNLocator, MultipleLocator

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc="lower right")

# Remove ALL y ticks + y tick labels
ax.yaxis.set_major_locator(NullLocator())

# Remove ONLY x tick labels (keep tick marks)
ax.xaxis.set_major_formatter(NullFormatter())

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc="lower right")

# At most 5 major ticks on x-axis
ax.xaxis.set_major_locator(MaxNLocator(5))

# Tick every 10 units on x-axis (0, 10, 20, 30, 40, 50)
ax.xaxis.set_major_locator(MultipleLocator(10))

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
